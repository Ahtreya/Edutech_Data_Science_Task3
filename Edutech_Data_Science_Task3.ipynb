```python
import pandas as pd
df=pd.read_csv('/content/Iris.csv')
print(df)
```

```python
# Remove Id column
df = df.drop('Id', axis=1)
print(df)
```

```python
# 1. Generate Summary Statistics
print("---- Summary Statistics ----")
summary_stats=df.describe()
print(summary_stats)
```

```python
# 2. Visualize distributions
print("---- Visualize Distribution ----")
import matplotlib.pyplot as plt
import seaborn as sns
plt.figure(figsize=(12,8))
plt.subplot(2,2,1)
sns.histplot(df['SepalLengthCm'],kde=True)
plt.subplot(2,2,2)
sns.histplot(df['SepalWidthCm'],kde=True)
plt.subplot(2,2,3)
sns.histplot(df['PetalLengthCm'],kde=True)
plt.subplot(2,2,4)
sns.histplot(df['PetalWidthCm'],kde=True)
plt.show()

```

```python
# 3. Create correlation heatmap
print("---- Correlation Heatmap ----")
import matplotlib.pyplot as plt
import seaborn as sns
plt.figure(figsize=(12,8))
sns.heatmap(df.drop('Species', axis=1).corr(), annot=True, cmap='Blues')
plt.show()
```

```python
# 4 & 5.  Detect patterns and Identify Trends
print("---- Detect Patterns and Identify Trends ----")
pair_plot=sns.pairplot(df,hue='Species',palette='inferno',markers='o',diag_kind='kde')
plt.show()
```
